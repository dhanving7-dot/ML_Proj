```python
plt.figure(figsize=(6,4))

sns.countplot(x="Class", data=df)

plt.title("Legitimate vs Fraudulent Transactions")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
class_counts = df["Class"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    class_counts,
    labels=["Legitimate", "Fraud"],
    autopct="%1.2f%%"
)

plt.title("Transaction Class Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Amount"], bins=50)

plt.title("Distribution of Transaction Amount")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Class", y="Amount", data=df)

plt.title("Transaction Amount by Class")
plt.xlabel("Class")
plt.ylabel("Transaction Amount")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Class",
    y="Amount",
    data=df,
    showfliers=False
)

plt.title("Transaction Amount by Class")
plt.xlabel("Class")
plt.ylabel("Transaction Amount")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df["Time"], bins=100)

plt.title("Transaction Time Distribution")
plt.xlabel("Time")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
fraud_df = df[df["Class"] == 1]

plt.figure(figsize=(10,5))

plt.scatter(
    fraud_df["Time"],
    fraud_df["Amount"],
    alpha=0.5
)

plt.title("Fraudulent Transactions Over Time")
plt.xlabel("Time")
plt.ylabel("Transaction Amount")

plt.show()
```

```python
print("Average legitimate transaction:",
      df[df["Class"] == 0]["Amount"].mean())

print("Average fraudulent transaction:",
      df[df["Class"] == 1]["Amount"].mean())

print("\nMedian legitimate transaction:",
      df[df["Class"] == 0]["Amount"].median())

print("Median fraudulent transaction:",
      df[df["Class"] == 1]["Amount"].median())
```

```python
plt.figure(figsize=(16,12))

correlation = df.corr()

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.show()
```

```python
correlation_with_class = df.corr()["Class"].sort_values()

print(correlation_with_class)
```

```python
plt.figure(figsize=(8,10))

correlation_with_class.drop("Class").plot(kind="barh")

plt.title("Feature Correlation with Fraud")
plt.xlabel("Correlation with Class")

plt.show()
```

```python
features = ["V1", "V2", "V3", "V4"]

for feature in features:

    plt.figure(figsize=(8,4))

    sns.kdeplot(
        data=df,
        x=feature,
        hue="Class",
        fill=True
    )

    plt.title(f"{feature} Distribution by Class")

    plt.show()
```

```python
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Amount"] < lower_bound) |
    (df["Amount"] > upper_bound)
]

print("Number of Amount outliers:", len(outliers))
```

```python
print("========== CREDIT CARD FRAUD DETECTION - EDA SUMMARY ==========")

print("Total transactions:", len(df))

print("Total columns:", df.shape[1])

print("\nClass Distribution:")
print(df["Class"].value_counts())

print("\nMissing Values:",
      df.isnull().sum().sum())

print("\nDuplicate Rows:",
      df.duplicated().sum())

fraud_percentage = (df["Class"].sum() / len(df)) * 100

print("\nFraud Percentage:",
      round(fraud_percentage, 4), "%")

print("\nAverage Transaction Amount:",
      round(df["Amount"].mean(), 2))

print("\nMedian Transaction Amount:",
      round(df["Amount"].median(), 2))

print("\n========== EDA COMPLETED ==========")
```
